**<font size="8" color="red">※cluster</font>**
## cluster

    * cluster(클러스터)는 “무리”, “집단”, “군집”이라는 뜻
    * 머신러닝에서는 특징이 비슷한 데이터끼리 모아 놓은 그룹
    * cluster라는 단어 자체가 특정 함수 하나를 뜻하는 것은 아니다.
    * **각 그룹이 cluster이고, 비슷한 데이터를 그룹으로 나누는 과정이 clustering(클러스터링, 군집화)**
    * 일반적으로 군집화는 정답 라벨 없이 데이터의 특징을 바탕으로 그룹을 찾는 비지도학습에 해당
- 2. from sklearn.cluster import KMeans에서의 의미
| 코드        | 의미                       |
| --------- | ------------------------ |
| `from`    | ~에서                      |
| `sklearn` | 머신러닝 라이브러리인 scikit-learn |
| `cluster` | 군집화 알고리즘이 들어 있는 모듈       |
| `import`  | 가져오기                     |
| `KMeans`  | K-평균 군집화를 수행하는 클래스       |

- 3. 코드에서 자주 보이는 관련 용어
    * ex.)model = KMeans(n_clusters=2, random_state=3, n_init=10)
        * 여기서 n_clusters=2는 데이터를 2개의 군집으로 나누겠다는 설정입니다.
            * n : 개수(number)를 나타내는 표현
            * clusters : 군집들
            * n_clusters : 군집의 개수
- 4. 함수·메서드·속성을 구분하면 더 이해하기 쉽습니다
| 표현                       | 종류        | 역할                 |
| ------------------------ | --------- | ------------------ |
| `sklearn.cluster`        | 모듈        | 군집화 도구를 모아 놓은 곳    |
| `KMeans`                 | 클래스       | 군집화 모델을 만드는 설계도    |
| `n_clusters`             | 매개변수      | 만들 군집의 개수를 설정      |
| `model.fit_predict(X)`   | 메서드 호출    | 군집을 학습하고 군집 번호를 반환 |
| `cluster_labels`         | 사용자 정의 변수 | 반환받은 군집 번호를 저장     |
| `model.cluster_centers_` | 속성        | 학습한 군집 중심 좌표를 제공   |

- NIST에 연결하면, 분류는 정답 숫자 0~9를 배우는 작업이고, 군집화는 정답을 알려 주지 않고 이미지의 특징이 비슷한 것끼리 묶는 작업입니다. 
- 따라서 군집화 결과의 ‘0번 군집’이 반드시 손글씨 숫자 ‘0’을 의미하지는 않습니다.

**<font size="8" color="red">※classification</font>**
## classification

- classification(클래시피케이션)은 ‘분류’라는 뜻
- 머신러닝에서는 데이터가 미리 정해진 범주 중 어디에 속하는지 예측하는 작업

| 용어               | 뜻               | MNIST 예시             |
| ---------------- | --------------- | -------------------- |
| `classification` | 분류하는 작업         | 이미지를 숫자 0~9 중 하나로 분류 |
| `class`          | 분류할 범주          | 숫자 0, 1, 2, …, 9     |
| `label`          | 데이터에 붙어 있는 정답 값 | 특정 이미지의 정답 `7`       |
| `classifier`     | 분류를 수행하는 모델     | 학습한 숫자 판별 모델         |
| `feature`        | 예측에 사용하는 입력 특징  | 이미지의 픽셀 값            |



- cluster와의 차이


| 구분        | 분류 `classification` | 군집화 `clustering`   |
| --------- | ------------------- | ------------------ |
| 학습할 때 정답  | 일반적으로 정답 라벨 제공      | 정답 라벨 없이 학습        |
| 목적        | 미리 정한 범주를 예측        | 비슷한 데이터끼리 그룹을 발견   |
| 대표 학습 방식  | 지도학습                | 비지도학습              |
| MNIST 예시  | “이 이미지는 숫자 7이다”     | “이 이미지들은 모양이 비슷하다” |
| 결과 번호의 의미 | 정해진 클래스에 대응         | 군집을 구별하기 위한 번호     |


**<font size="8" color="red">※LogisticRegression</font>**
## LogisticRegression

- LogisticRegression(로지스틱 회귀)은 데이터가 특정 클래스에 속할 확률을 계산하여 분류하는 머신러닝 모델
    * 이름에 회귀(Regression)가 있지만, 일반적인 용도는 분류(Classification)
    * 이진 분류에서는 보통 기준값을 0.5로 설정하여 분류합니다.
        * 합격 확률이 0.5보다 크면 → 합격으로 예측
        * 합격 확률이 0.5보다 작으면 → 불합격으로 예측
        * 0.5인 경우의 처리는 구현에 따라 달라질 수 있으며, 분류 기준은 목적에 따라 조정할 수도 있습니다.

- 2. 확률은 어떻게 계산하나요?        
    * 먼저 입력값에 가중치를 곱하고 절편을 더합니다.
        * z=wx+b
        | 기호   | 의미                     |
        | ----- | ---------------------- |
        | \(x\) | 입력값: 공부 시간             |
        | \(w\) | 가중치: 공부 시간이 예측에 미치는 영향 |
        | \(b\) | 절편: 입력과 무관하게 더해지는 값    |
        | \(z\) | 계산된 점수                 |

    * 이 점수는 음수나 1보다 큰 값도 될 수 있습니다. 따라서 시그모이드 함수를 이용해 0과 1 사이의 값으로 변환
    
- 3. 코드에서의 의미
    * ex)from sklearn.linear_model import LogisticRegression
        * sklearn : scikit-learn 머신러닝 라이브러리
        * linear_model : 선형 모델 관련 도구가 들어 있는 모듈
        * LogisticRegression : 로지스틱 회귀 모델을 만드는 클래스
        
- LogisticRegression은 클래스이고, 뒤에 ()를 붙이면 모델 객체가 만들어집니다. model은 그 객체를 저장하는 변수
- | 메서드               | 역할            | 이진 분류의 결과 예시  |
| ----------------- | ------------- | ------------- |
| `fit()`           | 학습 데이터로 모델 학습 | 가중치와 절편 등을 학습 |
| `predict()`       | 클래스 반환        | `1`           |
| `predict_proba()` | 클래스별 예측 확률 반환 | `[0.2, 0.8]`  |

- 로지스틱 회귀는 이진 분류뿐 아니라 MNIST처럼 여러 클래스가 있는 분류에도 사용할 수 있습
- 다중 클래스에서는 모델 구성에 따라 소프트맥스 등을 사용해 클래스별 확률을 계산
