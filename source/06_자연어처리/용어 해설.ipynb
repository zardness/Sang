# 1. attention

**<font size="6" color="red">※attention</font>**

- “입력 데이터 중에서 지금 중요한 부분에 더 집중하는 방법”
- Attention은 각 입력에 중요도(Attention Weight)를 계산
- Attention Distribution(어텐션 분포) : 각 입력 요소에 얼마나 집중할지를 확률처럼 나타낸 값들의 분포
- Attention의 핵심 용어
    * Query (Q) : 내가 지금 찾고 싶은 정보
    * Key (K)   : 각 정보가 무엇인지 나타내는 특징
    * Value (V) : 실제 가지고 있는 정보
    * Attention Score : Query와 Key가 얼마나 관련 있는지 계산한 원래 점수
    * Attention Weight : 정규화된 중요도
    * Attention Distribution : 여러 Attention Weight를 모아놓은 전체 분포
                               현재 모델이 입력의 어디를 얼마나 보고 있는가"를 숫자로 나타낸 분포
                               
        * Query
              ↓
            각 Key와 비교
              ↓
            Attention Score 계산
              ↓
            Softmax
              ↓
            Attention Weight
              ↓
            Value에 가중치 적용
              ↓
            중요한 정보들을 합침
              ↓
            Attention 결과
            
    * Attention과 Softmax는 밀접한 관계
        * Q와 K 비교
               ↓
            관련성 점수 계산
               ↓
            Softmax로 중요도 계산
               ↓
            중요도 × V
               ↓
            필요한 정보에 집중

    * 여기서 더 발전한 것이 Transformer입니다. 
      Transformer는 RNN 없이 Attention, 특히 Self-Attention을 핵심으로 사용
    * 현재 필요한 정보를 찾고, 중요한 부분에 더 큰 가중치를 주어 집중하는 딥러닝 기법
    * 학습 순서는 RNN → Seq2Seq → Attention → Q·K·V → Self-Attention → Transformer 순으로 
      공부하면 연결해서 이해하기 좋습니다.


# 2. Transforme

**<font size="6" color="red">※Transformer</font>**
- Transformer(트랜스포머)는 자연어 처리에서 매우 중요한 딥러닝 모델 구조로, 
  핵심은 Attention을 이용해서 문장 속 단어들 사이의 관계를 파악하는 것.
  
- Transformer의 핵심: Self-Attention

- Self-Attention에서는 각 단어로부터 세 가지 정보를 만듭니다
    * Query : 내가 찾고 싶은 정보
    * Key   : 내가 어떤 정보인지 나타내는 특징
    * Value : 내가 실제로 가지고 있는 정보
    * 계산 순서
        * Q와 K 비교
               ↓
            QKᵀ
               ↓
            Attention Score
               ↓
            √dk 로 나누기
               ↓
            Softmax
               ↓
            Attention Distribution
               ↓
            Value(V)에 적용
               ↓
            Attention 결과
            
            * Softmax, Attention Distribution이 여기서 연결
            
- Multi-Head Attention : 한 번만 계산하지 않고 여러 개의 Attention을 동시에 사용

- Transformer 전체 구조
    *     입력 문장
    *        ↓
    *     Embedding
    *        ↓
    *     Positional Encoding
    *        ↓
    *    ┌───────────────┐
    *    │    Encoder    │
    *    │               │
    *    │ Multi-Head    │
    *    │ Self-Attention│
    *    │      ↓        │
    *    │ Feed Forward  │
    *    └───────────────┘
    *           ↓
    *    Encoder 결과
    *           ↓
    *    ┌───────────────┐
    *    │    Decoder    │
    *    │               │
    *    │ Masked        │
    *    │ Self-Attention│
    *    │      ↓        │
    *    │ Cross         │
    *    │ Attention     │
    *    │      ↓        │
    *    │ Feed Forward  │
    *    └───────────────┘
    *           ↓
    *         Linear
    *           ↓
    *        Softmax
    *           ↓
    *    다음 단어 확률
        
- Transformer의 핵심 = Self-Attention
- Self-Attention의 핵심 : Q와 K로 중요도를 계산하고 → Softmax로 Attention Distribution을 만든 다음 → 
                         그 중요도를 V에 적용한다



      * DNN
         ↓
        RNN
         ↓
        Seq2Seq
         ↓
        Attention
         ↓
        Attention Score
         ↓
        Softmax
         ↓
        Attention Distribution
         ↓
        Self-Attention
         ↓
        Multi-Head Attention
         ↓
        Transformer
     

# 3. confusion

**<font size="6" color="red">※confusion</font>**
- Confusion Matrix(혼동 행렬)를 의미
- Confusion Matrix = 분류 모델이 실제 정답과 예측값을 얼마나 맞추고 틀렸는지 표로 나타낸 것

- 	예측: 정상(N)	예측: 스팸(P)
    실제: 정상(N)	TN	FP
    실제: 스팸(P)	FN	TP
        * TP (True Positive) : 실제로 스팸이고 모델도 스팸이라고 예측 → 정답
        * TN (True Negative) : 실제로 정상이고 모델도 정상이라고 예측 → 정답
        * FP (False Positive) : 실제로는 정상인데 모델이 스팸이라고 예측 → 오답
        * FN (False Negative) : 실제로는 스팸인데 모델이 정상이라고 예측 → 오답
        
        * True  → 예측이 맞았다
          False → 예측이 틀렸다

          Positive → 모델이 양성이라고 판단
          Negative → 모델이 음성이라고 판단
              
        * EX)      
              * TP → Positive라고 예측했고 맞음
                FP → Positive라고 예측했는데 틀림

                TN → Negative라고 예측했고 맞음
                FN → Negative라고 예측했는데 틀림
