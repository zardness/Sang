### pipeline

- 여러 작업 단계를 하나의 흐름으로 연결해서 처리하는 것
- 복잡한 자연어 처리 모델을 쉽게 사용할 수 있도록 전처리 → 모델 실행 → 후처리를 하나로 묶어주는 함수
- pipeline("text-classification") : "text-classification"은 어떤 NLP 작업을 할 것인지 지정

    * "I love this movie."
        ↓
        ① Tokenizer
        문장을 모델이 처리할 수 있도록 변환
                ↓
        ② Model
        Transformer 모델이 계산
                ↓
        ③ 예측값 계산
                ↓
        ④ 후처리
        사람이 이해하기 쉬운 형태로 변환
                ↓
        POSITIVE / score
        
- "text-classification" : 문장 분류
- "sentiment-analysis"  : 감성 분석
- zero-shot-classification"	: 제시한 후보 라벨로 분류
- "question-answering"	: 질문에 답하기
- "summarization"	: 문서 요약
- "translation"	 : 번역
- "text-generation"	: 문장 생성

    * pipeline
           │
           ├─ 작업 지정
           │   └─ zero-shot-classification
           │
           ├─ Tokenizer
           │
           ├─ Transformer Model
           │
           └─ 결과 후처리
           
- pipeline() = Hugging Face에서 복잡한 AI 모델 사용 과정을 하나로 묶어서 간단하게 실행할 수 있게 해주는 함수

**<font size="8" color="red">※client</font>**
## client

- 프로그래밍에서 서버(Server)에 요청을 보내고 결과를 받아 사용하는 쪽을 의미
   * Client = 서비스를 요청하는 쪽
   * Server = 요청을 받아 처리하고 결과를 보내주는 쪽
- client = OpenAI()
    * 여기서 client는 OpenAI 서비스에 요청을 보내기 위해 만든 객체를 저장하는 변수 이름
    * client가 Python의 특별한 예약어는 아니라는 것

**<font size="8" color="red">※response</font>**
## response

- 프로그래밍에서 응답, 반환된 결과라는 뜻
    * Request = 요청
      Response = 요청에 대한 응답
- response = ...
    * 서버가 보내준 응답 결과를 response라는 변수에 저장한다는 뜻

- client와 response 연결
    * client
          │
          │ 요청을 보냄
          ↓
        Server
          │
          │ 결과를 보냄
          ↓
        response
        
- response 안에는 단순한 문자열 하나만 들어있는 것이 아니라 여러 정보가 포함될 수 있슴
    * response
    *   ├─ 실제 AI 답변
    *   ├─ 모델 정보
    *   ├─ 토큰 사용량
    *   └─ 기타 응답 정보
- 실제 답변 텍스트를 얻으려면 사용하는 라이브러리에 따라 response 안의 특정 속성에 접근하게 됨
- response 역시 Python의 예약어가 아니다.
- 개발자가 정한 변수 이름
    * client   = 요청을 보내는 객체
      request  = 보내는 요청
      server   = 요청을 처리
      response = 처리 후 돌아온 응답

**<font size="8" color="red">※frequency_penalty</font>**
## frequency_penalty

- LLM에서 같은 단어나 표현이 반복되는 것을 얼마나 억제할지 조절하는 값
- 이미 여러 번 사용한 단어를 또 사용하는 것에 페널티를 주는 설정
- 값의 의미
    * frequency_penalty 낮음
                ↓
        반복 표현을 크게 신경 쓰지 않음

        frequency_penalty 높음
                ↓
        이미 자주 사용한 표현을 피하려고 함
        
- 왜 frequency라는 이름인가? =>frequency는 빈도, 횟수라는 뜻이고 penalty는 벌점이라는 뜻
    * frequency + penalty
       빈도         벌점

    = 많이 등장한 것에 벌점을 줌
    
- temperature와 차이
    * temperature
        → 답변의 무작위성·다양성을 조절

        frequency_penalty
        → 같은 표현의 반복을 억제
        
- frequency_penalty = 이미 자주 등장한 토큰에 페널티를 줘서 반복을 줄이는 설정

**<font size="8" color="red">※few shot</font>**
## few shot
- LLM에게 작업을 시킬 때 몇 개의 예시(Example)를 먼저 보여준 뒤, 그 예시를 참고해서 새로운 문제를 해결하게 하는 방법

- Few-shot = “예시 몇 개 보여줄 테니까, 이 방식대로 해봐.”
    * Zero-shot : 예시 없이 바로 질문
      One-shot  : 예시 1개 제공
      Few-shot  : 예시 여러 개 제공
      
- 중요한 점
    Few-shot은 일반적으로 모델 자체를 다시 학습시키는 것(training)과는 다름
        * Training
            → 모델의 가중치(Weight)를 실제로 변경
         
        * Few-shot Prompting
            → Weight를 변경하지 않음
            → Prompt 안에 예시를 제공
            → 모델이 예시의 패턴을 참고해서 답변
- Few-shot = LLM에게 몇 개의 예제를 프롬프트로 제공하고, 
            그 예제의 형식이나 패턴을 참고해서 새로운 문제를 해결하게 하는 방법
