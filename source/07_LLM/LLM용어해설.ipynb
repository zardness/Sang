### zero-shot-classification

  사용자가 원하는 후보 카테고리를 제시하면 그중 어떤 카테고리가 입력 문장과 가장 관련 있는지 판단하는 방식

**<font size="8" color="red">※pipeline</font>**

### pipeline

- 여러 작업 단계를 하나의 흐름으로 연결해서 처리하는 것
- 복잡한 자연어 처리 모델을 쉽게 사용할 수 있도록 전처리 → 모델 실행 → 후처리를 하나로 묶어주는 함수
- pipeline("text-classification") : "text-classification"은 어떤 NLP 작업을 할 것인지 지정

    * "I love this movie."
        ↓
        ① Tokenizer
        문장을 모델이 처리할 수 있도록 변환
                ↓
        ② Model
        Transformer 모델이 계산
                ↓
        ③ 예측값 계산
                ↓
        ④ 후처리
        사람이 이해하기 쉬운 형태로 변환
                ↓
        POSITIVE / score
        
- "text-classification" : 문장 분류
- "sentiment-analysis"  : 감성 분석
- zero-shot-classification"	: 제시한 후보 라벨로 분류
- "question-answering"	: 질문에 답하기
- "summarization"	: 문서 요약
- "translation"	 : 번역
- "text-generation"	: 문장 생성

    * pipeline
           │
           ├─ 작업 지정
           │   └─ zero-shot-classification
           │
           ├─ Tokenizer
           │
           ├─ Transformer Model
           │
           └─ 결과 후처리
           
- pipeline() = Hugging Face에서 복잡한 AI 모델 사용 과정을 하나로 묶어서 간단하게 실행할 수 있게 해주는 함수

**<font size="8" color="red">※invoke</font>**

- “호출하다, 실행시키다”라는 뜻
- 함수, 메서드, 모델, 체인 등을 호출해서 실제로 실행

    * model.invoke("안녕하세요")

                ↓

        model에게
        "안녕하세요"를 입력해서

                ↓

        모델을 실행하고

                ↓

        결과를 반환
        
- invoke() = “이 객체에 입력값을 넣어서 지금 실행해!”

**<font size="8" color="red">※context(컨텍스트)=ctx</font>**
- 현재 작업을 처리하는 데 필요한 주변 정보, 상태, 문맥
- 예약어나 정해진 함수가 아니라 개발자가 임의로 정하는 변수 이름
- LLM 분야에서는 ctx가 특히 Context(문맥)를 의미하는 경우가 많습니다.

**<font size="8" color="red">※Template(컨텍스트)=ctx</font>**
- 프로그래밍에서 미리 만들어 놓은 틀 또는 양식
- LangChain / LLM에서는 보통 Prompt Template(프롬프트 템플릿)을 의미
- EX)
- result = template.invoke({
    "topic": "Transformer"})
     print(result)
     
         * Template
               ↓
            질문의 형식을 만든다

            LLM
               ↓
            그 질문을 받아 답변을 생성한다
            
- 반복해서 사용할 문장이나 프롬프트의 기본 틀

**<font size="8" color="red">※parser</font>**
- 프로그래밍에서 데이터를 분석해서 원하는 형태로 변환하는 역할
- LLM이 생성한 답변을 우리가 사용하기 편한 형태로 변환하는 것을 의미
- **Parser = LLM의 출력 결과를 분석해서 프로그램에서 사용하기 좋은 형태로 변환하는 도구**

**<font size="8" color="red">※StrOutputParser</font>**
- LLM의 출력을 문자열(String) 형태로 가져오는 Parser

**<font size="8" color="red">※temperature</font>**
- 모델이 답변을 생성할 때 얼마나 다양하고 무작위성 있게 단어를 선택할지 조절하는 값
- temperature가 낮으면 안정적·일관된 답변, 높으면 다양하고 창의적인 답변
- 창작, 아이디어 생성, 브레인스토밍처럼 다양성이 필요한 작업에서 유용
- 
        temperature  	특징	                  적합한 작업
            낮음	   안정적, 일관적	           코드, 분석, 정보 추출
            중간	   안정성과 다양성의 균형	         일반 대화
            높음	   다양성 증가	             창작, 아이디어 생성
